# LEVEL 2 – TASK 1
## Table Booking and Online Delivery Analysis

### Step 1: Importing Required Libraries

Python libraries such as Pandas, Matplotlib, and Seaborn are imported for data analysis and visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Step 2: Uploading Dataset

The restaurant dataset is uploaded into Google Colab for analysis.

```python
from google.colab import files
uploaded = files.upload()
```

### Step 3: Loading Dataset

The uploaded CSV file is loaded into a Pandas DataFrame.

```python
df = pd.read_csv('Dataset .csv')

df.head()
```

### Step 4: Displaying Column Names

This step displays all column names available in the dataset.

```python
print(df.columns)
```

### Step 5: Table Booking Percentage Analysis

This analysis calculates the percentage of restaurants that provide table booking facilities.

```python
table_booking_counts = df['Has Table booking'].value_counts()

table_booking_percentage = (
    table_booking_counts / len(df)
) * 100

print(table_booking_percentage)
```

### Step 6: Visualization of Table Booking Availability

A pie chart is created to visualize the percentage of restaurants with and without table booking services.

```python
plt.figure(figsize=(6,6))

plt.pie(
    table_booking_counts,
    labels=table_booking_counts.index,
    autopct='%1.1f%%'
)

plt.title('Table Booking Availability')

plt.show()
```

### Step 7: Online Delivery Percentage Analysis

This step calculates the percentage of restaurants offering online delivery services.

```python
online_delivery_counts = df['Has Online delivery'].value_counts()

online_delivery_percentage = (
    online_delivery_counts / len(df)
) * 100

print(online_delivery_percentage)
```

### Step 8: Visualization of Online Delivery Availability

A pie chart is generated to display restaurants with and without online delivery services.

```python
plt.figure(figsize=(6,6))

plt.pie(
    online_delivery_counts,
    labels=online_delivery_counts.index,
    autopct='%1.1f%%'
)

plt.title('Online Delivery Availability')

plt.show()
```

### Step 9: Average Rating Comparison for Table Booking

This analysis compares the average aggregate ratings of restaurants that provide table booking and those that do not.

```python
avg_rating_booking = df.groupby(
    'Has Table booking'
)['Aggregate rating'].mean()

print(avg_rating_booking)
```

### Step 10: Visualization of Ratings by Table Booking

A bar chart is created to compare restaurant ratings based on table booking availability.

```python
plt.figure(figsize=(6,4))

sns.barplot(
    x=avg_rating_booking.index,
    y=avg_rating_booking.values
)

plt.title('Average Ratings by Table Booking')
plt.xlabel('Table Booking')
plt.ylabel('Average Rating')

plt.show()
```

### Step 11: Online Delivery Across Different Price Ranges

This analysis identifies how online delivery services vary across different restaurant price ranges.

```python
delivery_price = pd.crosstab(
    df['Price range'],
    df['Has Online delivery']
)

print(delivery_price)
```

### Step 12: Visualization of Online Delivery and Price Range

A stacked bar chart is created to analyze online delivery availability across price categories.

```python
delivery_price.plot(
    kind='bar',
    stacked=True,
    figsize=(8,5)
)

plt.title('Online Delivery Across Price Ranges')
plt.xlabel('Price Range')
plt.ylabel('Number of Restaurants')

plt.show()
```

# Conclusion

In Level 2 Task 1, the availability of table booking and online delivery services was analyzed. Percentage analysis and visualizations helped identify restaurants providing these facilities. Average ratings were compared between restaurants with and without table booking services. Additionally, the relationship between online delivery and price range was explored using data visualization techniques.

## Price Range Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Step 2: Uploading Dataset

The restaurant dataset is uploaded into Google Colab for further analysis.

```python
from google.colab import files
uploaded = files.upload()
```

### Step 3: Loading Dataset

The CSV dataset is loaded into a Pandas DataFrame for analysis.

```python
df = pd.read_csv('Dataset .csv')

df.head()
```

### Step 4: Checking Dataset Columns

This step displays all columns available in the dataset.

```python
print(df.columns)
```

### Step 5: Price Range Distribution Analysis

This analysis identifies the most common price range among all restaurants in the dataset.

```python
price_range_counts = df['Price range'].value_counts()

print(price_range_counts)
```

### Step 6: Visualization of Price Range Distribution

A bar chart is created to visualize restaurant counts across different price ranges.

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x=price_range_counts.index,
    y=price_range_counts.values
)

plt.title('Distribution of Price Ranges')
plt.xlabel('Price Range')
plt.ylabel('Number of Restaurants')

plt.show()
```

### Step 7: Average Ratings by Price Range

This analysis calculates the average aggregate rating for each restaurant price range.

```python
avg_rating_price = df.groupby(
    'Price range'
)['Aggregate rating'].mean()

print(avg_rating_price)
```

### Step 8: Visualization of Average Ratings by Price Range

A bar graph is created to compare average ratings across different price ranges.

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x=avg_rating_price.index,
    y=avg_rating_price.values
)

plt.title('Average Rating by Price Range')
plt.xlabel('Price Range')
plt.ylabel('Average Rating')

plt.show()
```

### Step 9: Rating Color Analysis

This step identifies the colors associated with restaurant ratings and determines which color represents the highest average rating.

```python
rating_color = df.groupby(
    'Rating color'
)['Aggregate rating'].mean().sort_values(ascending=False)

print(rating_color)
```

### Step 10: Visualization of Rating Colors

A bar chart is used to compare average ratings represented by different rating colors.

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x=rating_color.index,
    y=rating_color.values
)

plt.title('Average Rating by Rating Color')
plt.xlabel('Rating Color')
plt.ylabel('Average Rating')

plt.show()
```

### Step 11: Identifying Highest Rated Color

This step determines the rating color associated with the highest average restaurant rating.

```python
highest_color = rating_color.idxmax()

print("Color with Highest Average Rating:")
print(highest_color)
```

# Conclusion

In Level 2 Task 2, price range analysis was performed on the restaurant dataset. The most common price range was identified, and average ratings were calculated for each price category. Rating colors were also analyzed to determine which color represented the highest average rating. Visualizations helped understand the relationship between restaurant pricing and customer ratings.

# LEVEL 2 – TASK 3
## Feature Engineering

### Step 1: Importing Required Libraries

Python libraries such as Pandas, NumPy, Matplotlib, and Seaborn are imported for feature engineering and data analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Step 2: Uploading Dataset

The restaurant dataset is uploaded into Google Colab for feature engineering analysis.

```python
from google.colab import files
uploaded = files.upload()
```

### Step 3: Loading Dataset

The CSV file is loaded into a Pandas DataFrame for further processing.

```python
df = pd.read_csv('Dataset .csv')

df.head()
```

### Step 4: Displaying Dataset Columns

This step displays all available columns in the dataset.

```python
print(df.columns)
```

### Step 5: Creating Restaurant Name Length Feature

A new feature is created to calculate the length of each restaurant name.

```python
df['Restaurant Name Length'] = df['Restaurant Name'].apply(len)

print(df[['Restaurant Name', 'Restaurant Name Length']].head())
```

### Step 6: Creating Address Length Feature

A new feature is generated to calculate the length of restaurant addresses.

```python
df['Address Length'] = df['Address'].apply(len)

print(df[['Address', 'Address Length']].head())
```

### Step 7: Encoding Table Booking Feature

The categorical column 'Has Table booking' is converted into numerical format using binary encoding.

```python
df['Has Table Booking Binary'] = df[
    'Has Table booking'
].map({
    'Yes': 1,
    'No': 0
})

print(df[['Has Table booking', 'Has Table Booking Binary']].head())
```

### Step 8: Encoding Online Delivery Feature

The 'Has Online delivery' column is converted into numerical values for machine learning purposes.

```python
df['Has Online Delivery Binary'] = df[
    'Has Online delivery'
].map({
    'Yes': 1,
    'No': 0
})

print(df[['Has Online delivery', 'Has Online Delivery Binary']].head())
```

### Step 9: Creating Number of Cuisines Feature

This feature calculates how many cuisines are offered by each restaurant.

```python
df['Number of Cuisines'] = df[
    'Cuisines'
].apply(
    lambda x: len(str(x).split(','))
)

print(df[['Cuisines', 'Number of Cuisines']].head())
```

### Step 10: Statistical Summary of New Features

Descriptive statistics are generated for the newly engineered features.### Step 10: Statistical Summary of New Features

Descriptive statistics are generated for the newly engineered features.

```python
print(df[[
    'Restaurant Name Length',
    'Address Length',
    'Has Table Booking Binary',
    'Has Online Delivery Binary',
    'Number of Cuisines'
]].describe())
```

### Step 11: Visualization of Restaurant Name Length

A histogram is created to visualize the distribution of restaurant name lengths.

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['Restaurant Name Length'],
    bins=20,
    kde=True
)

plt.title('Distribution of Restaurant Name Length')
plt.xlabel('Restaurant Name Length')
plt.ylabel('Count')

plt.show()
```

### Step 12: Visualization of Number of Cuisines

This graph displays how many cuisines are offered by restaurants in the dataset.

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x=df['Number of Cuisines']
)

plt.title('Number of Cuisines Offered')
plt.xlabel('Number of Cuisines')
plt.ylabel('Restaurant Count')

plt.show()
```

### Step 13: Correlation Analysis of Engineered Features

Correlation analysis is performed to identify relationships among newly created numerical features.

```python
correlation = df[[
    'Restaurant Name Length',
    'Address Length',
    'Has Table Booking Binary',
    'Has Online Delivery Binary',
    'Number of Cuisines',
    'Aggregate rating'
]].corr()

print(correlation)
```

### Step 14: Heatmap of Engineered Features

A heatmap is generated to visually represent correlations among engineered features and aggregate ratings.

```python
plt.figure(figsize=(10,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title('Feature Correlation Heatmap')

plt.show()
```

# Conclusion

In Level 2 Task 3, feature engineering techniques were applied to the restaurant dataset. New features such as restaurant name length, address length, number of cuisines, and binary encoded variables for table booking and online delivery were created. Statistical analysis and visualizations helped understand the relationships between engineered features and restaurant ratings. This task improved understanding of feature creation and preprocessing techniques used in data science projects.
